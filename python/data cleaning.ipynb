```python
df["Shipping Address"] = (
    df["Shipping Address"]
    .str.replace("\n", " ", regex=False)
    .str.strip()
)
```

```python
df["Billing Address"] = (
    df["Billing Address"]
    .str.replace("\n", " ", regex=False)
    .str.strip()
)
```

```python
text_columns = [
    "Payment Method",
    "Product Category",
    "Customer Location",
    "Device Used"
]

for col in text_columns:
    df[col] = df[col].str.strip().str.lower()
```

```python
for col in text_columns:
    print("\n", col)
    print(df[col].unique())
```

```python
import ipaddress

def valid_ip(ip):
    try:
        ipaddress.ip_address(ip)
        return True
    except:
        return False

df["Valid IP"] = df["IP Address"].apply(valid_ip)

df["Valid IP"].value_counts()
```

```python
df.drop(columns=["Valid IP"], inplace=True)
```
